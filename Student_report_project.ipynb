```python
import json
import os
```

```python
class Student:
    _id_counter = 1  # auto-increment ID

    def __init__(self, name, subjects=None, student_id=None):
        if student_id is None:
            self.id = Student._id_counter
            Student._id_counter += 1
        else:
            self.id = student_id
            Student._id_counter = max(Student._id_counter, student_id + 1)

        self.name = name
        self.subjects = subjects if subjects else {}

    def add_subject(self, subject, score):
        self.subjects[subject] = score

    def calculate_average(self):
        if not self.subjects:
            return 0
        return sum(self.subjects.values()) / len(self.subjects)

    def get_grade(self):
        avg = self.calculate_average()
        if avg >= 90:
            return "A"
        elif avg >= 75:
            return "B"
        elif avg >= 50:
            return "C"
        else:
            return "Fail"

    def to_dict(self):
        return {"id": self.id, "name": self.name, "subjects": self.subjects}

    @classmethod
    def from_dict(cls, data):
        return cls(name=data["name"], subjects=data["subjects"], student_id=data["id"])
```

```python
class GradeManager:
    def __init__(self, filename="grades.json"):
        self.students = []
        self.filename = filename
        self.load_from_file()

    def add_student(self, name, subjects):
        student = Student(name, subjects)
        self.students.append(student)
        print(f"✅ Student '{name}' added with ID {student.id}")

    def update_scores(self, student_id, subject, score):
        student = self.find_student(student_id)
        if student:
            student.add_subject(subject, score)
            print(f"✅ Updated {subject} score for {student.name}")
        else:
            print("❌ Student not found!")

    def view_report(self, student_id):
        student = self.find_student(student_id)
        if student:
            print(f"\n📘 Report Card for {student.name} (ID: {student.id})")
            for subject, score in student.subjects.items():
                print(f"   {subject}: {score}")
            avg = student.calculate_average()
            grade = student.get_grade()
            print(f"   Average: {avg:.2f}")
            print(f"   Grade: {grade}\n")
        else:
            print("❌ Student not found!")

    def delete_student(self, student_id):
        student = self.find_student(student_id)
        if student:
            self.students.remove(student)
            print(f"🗑️ Student {student.name} deleted successfully")
        else:
            print("❌ Student not found!")

    def save_to_file(self):
        data = [student.to_dict() for student in self.students]
        with open(self.filename, "w") as f:
            json.dump(data, f, indent=2)
        print("💾 Data saved to file")

    def load_from_file(self):
        if os.path.exists(self.filename):
            with open(self.filename, "r") as f:
                data = json.load(f)
                self.students = [Student.from_dict(d) for d in data]
            print("📂 Data loaded from file")
        else:
            print("⚠️ No existing data found, starting fresh")

    def find_student(self, student_id):
        for student in self.students:
            if student.id == student_id:
                return student
        return None

    def list_all_students(self):
        if not self.students:
            print("⚠️ No students found")
        else:
            print("\n👩‍🎓 Student List:")
            for student in self.students:
                print(f"   ID: {student.id}, Name: {student.name}, Grade: {student.get_grade()}")

```

```python
def generate_sample_dataset(filename="grades.json"):
    sample_data = [
        {"id": 1, "name": "Alice Johnson", "subjects": {"Math": 88, "Science": 92, "English": 85}},
        {"id": 2, "name": "Brian Smith", "subjects": {"Math": 76, "Science": 81, "English": 69, "History": 74}},
        {"id": 3, "name": "Catherine Lee", "subjects": {"Math": 95, "Science": 98, "English": 93, "Computer": 100}},
        {"id": 4, "name": "David Kim", "subjects": {"Math": 60, "Science": 55, "English": 65}}
    ]
    with open(filename, "w") as f:
        json.dump(sample_data, f, indent=2)
    print(f"✅ Sample dataset created in '{filename}'")
```

```python
def main():
    # Generate dataset automatically if file doesn't exist
    if not os.path.exists("grades.json"):
        generate_sample_dataset()

    manager = GradeManager()

    while True:
        print("\n===== Student Report Card Manager =====")
        print("1. Add Student")
        print("2. Update Scores")
        print("3. View Report")
        print("4. Delete Student")
        print("5. List All Students")
        print("6. Save Data")
        print("7. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            name = input("Enter student name: ")
            subjects = {}
            while True:
                subject = input("Enter subject (or 'done' to finish): ")
                if subject.lower() == "done":
                    break
                try:
                    score = int(input(f"Enter score for {subject}: "))
                except ValueError:
                    print("❌ Invalid score, must be a number")
                    continue
                subjects[subject] = score
            manager.add_student(name, subjects)

        elif choice == "2":
            try:
                student_id = int(input("Enter student ID: "))
                subject = input("Enter subject: ")
                score = int(input("Enter new score: "))
                manager.update_scores(student_id, subject, score)
            except ValueError:
                print("❌ Invalid input")

        elif choice == "3":
            try:
                student_id = int(input("Enter student ID: "))
                manager.view_report(student_id)
            except ValueError:
                print("❌ Invalid input")

        elif choice == "4":
            try:
                student_id = int(input("Enter student ID: "))
                manager.delete_student(student_id)
            except ValueError:
                print("❌ Invalid input")

        elif choice == "5":
            manager.list_all_students()

        elif choice == "6":
            manager.save_to_file()

        elif choice == "7":
            manager.save_to_file()
            print("👋 Exiting program...")
            break

        else:
            print("❌ Invalid choice, try again!")
```

```python
if __name__ == "__main__":
    main()
```
